# Task A

```python
import csv

# reader() function to load the data into memory and store it in a list

def load_csv_data(file_path):
    try:
        with open(file_path, mode='r') as file:
            reader = csv.reader(file)
            data = [row for row in reader]
        return data
    
    except FileNotFoundError:
        print(f"Error: The file '{file_path}' was not found.")
        return []

data = load_csv_data('lung_cancer_data.csv')
```

```python
# a1

def retrieve_demographic_info(data, patient_id):

    header, rows = data[0], data[1:]
    for row in rows:
        if row[header.index('Patient_ID')] == patient_id:
            return {"Age":row[header.index('Age')] ,
                    "Gender":row[header.index('Gender')],
                    "Smoking History":row[header.index('Smoking_History')],
                    "Ethnicity":row[header.index('Ethnicity')]}
    return None  

retrieve_demographic_info(data, patient_id="13117")
```

```python
# a2

def retrieve_medical_history(data, ethnicity):

    header, rows = data[0], data[1:]
    results = []
    for row in rows:
        if row[header.index('Ethnicity')]== ethnicity:
            results.append({
                "Family History": row[header.index('Family_History')],
                "Diabetes": row[header.index('Comorbidity_Diabetes')], 
                "Kidney Disease": row[header.index('Comorbidity_Kidney_Disease')], 
                "Haemoglobin Level": row[header.index('Haemoglobin_Level')]
            })
    
    return results
        
retrieve_medical_history(data ,"Hispanic")
```

```python
# a3

def retrieve_treatment_details(data, treatment):

    header, rows = data[0], data[1:]
    results = []
    for row in rows:
        if row[header.index('Treatment')]== treatment and int(row[header.index('Survival_Months')]) > 100:
            results.append({
                    "Age": row[header.index('Age')],
                    "Tumor Size": row[header.index('Tumor_Size_mm')],
                    "Tumor Location": row[header.index('Tumor_Location')],
                    "Tumor Stage": row[header.index('Stage')]
                }
            )
    return results
            
retrieve_treatment_details(data, treatment="Surgery")
```

```python
# a4

def retrieve_custom_info(data):
    header, rows = data[0], data[1:]
    results = []
    for row in rows:
        if float(row[header.index('White_Blood_Cell_Count')]) > 5:
            results.append({
                "Blood Pressure Systolic": row[header.index('Blood_Pressure_Systolic')],
                "Platelet Count": row[header.index('Platelet_Count')],
                "Albumin Level": row[header.index('Albumin_Level')]
            })
    return results

retrieve_custom_info(data)
```

# Task B

```python
import pandas as pd

df = pd.read_csv('lung_cancer_data.csv')
```

```python
# b1

def top_3_treatments(df, ethnicity):

    filtered_df = df[(df['Ethnicity'] == ethnicity) & (df['Survival_Months'] > 100)]
    treatment_counts = filtered_df.groupby('Treatment').size()
    top_treatments = treatment_counts.nlargest(3)
    return top_treatments

top_3_treatments(df, ethnicity="Hispanic")
```

```python
# b2

def average_wbc_counts(df, ethnicity, treatments):

    filtered_df = df[(df['Ethnicity'] == ethnicity) & (df['Treatment'].isin(treatments))]
    average_wbc = filtered_df.groupby('Treatment')['White_Blood_Cell_Count'].mean()
    return average_wbc

average_wbc_counts(df, ethnicity="Hispanic", treatments=["Surgery","Radiation Therapy"])
```

```python
# b3

def average_smoking_packs(df):
    filtered_df = df[(df['Blood_Pressure_Pulse'] > 90) & (df['Tumor_Size_mm'] < 15.0)]
    average_smoking = filtered_df.groupby(['Treatment', 'Tumor_Location'])['Smoking_Pack_Years'].mean().unstack()
    return average_smoking

average_smoking_packs(df)
```

```python
# b4

def average_survival_by_stage(df):

    filtered_df = df[(df['Comorbidity_Diabetes'] == "Yes")]
    average_survival = filtered_df.groupby(['Stage'])['Survival_Months'].mean()
    return average_survival

average_survival_by_stage(df)
```

# Task C

```python
import matplotlib.pyplot as plt
```

```python
# c1

def plot_treatment_proportions(df, ethnicity):

    filtered_df = df[df['Ethnicity'] == ethnicity]
    treatment_counts = filtered_df.groupby('Treatment').size()

    treatment_counts.plot(kind='bar')
    plt.title(f'Proportion of Cancer Treatments among {ethnicity}')
    plt.xlabel('Treatment')
    plt.ylabel('Count')
    plt.show()

# Example usage
plot_treatment_proportions(df, ethnicity="Hispanic")
```

```python
# c2

def plot_smoking_packs_by_stage(df):

    average_smoking_packs = df.groupby(['Ethnicity', 'Stage'])['Smoking_Pack_Years'].mean().unstack()
   
    average_smoking_packs.plot(kind='line')
    plt.title('Average Smoking Packs Consumption Across Different Cancer Stages by Ethnicity')
    plt.xlabel('Cancer Stage')
    plt.ylabel('Average Smoking Packs')
    plt.show()

# Example usage
plot_smoking_packs_by_stage(df)
```

```python
# c3 

def plot_blood_pressure_by_treatment(df):

    blood_pressures = ['Blood_Pressure_Systolic', 'Blood_Pressure_Diastolic', 'Blood_Pressure_Pulse']
    average_blood_pressure = df.groupby('Treatment')[blood_pressures].mean()

    average_blood_pressure.plot(kind='bar')
    plt.title('Average Blood Pressure Types Across Different Treatment Types')
    plt.xlabel('Treatment')
    plt.ylabel('Average Blood Pressure')
    plt.legend(title='Blood Pressure Type')
    plt.show()

# Example usage
plot_blood_pressure_by_treatment(df)
```

```python
# c4

def plot_treatment_effectiveness_by_age(df):

    df['Age_Group'] = df['Age'].apply(lambda x: round(x / 10) * 10)
    average_survival_months = df.groupby(['Treatment','Age_Group'])['Survival_Months'].mean().unstack()
   
    average_survival_months.plot(kind='line')
    plt.title('Average Survival Months by Treatment and Age Group')
    plt.xlabel('Treatment')
    plt.ylabel('Average Survival Months')
    plt.legend(title='Age Group')
    plt.show()

# Example usage
plot_treatment_effectiveness_by_age(df)
```
